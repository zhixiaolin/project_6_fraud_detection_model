```python
import pandas as pd
import numpy as np
from pandas import DataFrame

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn import preprocessing
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
path=r'C:\Users\aleci\Documents\Project 6 - Credit Card Fraud Detection\data\\'

df=pd.read_csv(path+'creditcard.csv')
```

```python
X=['V1', 'V2', 'V3', 'V4', 'V5', 'V6', 'V7', 'V8', 'V9', 'V10',
'V11','V12','V13','V14','V15','V16','V17','V18','V19','V20','V21','V22','V23','V24','V25','V26','V27','V28']

df=df[X].T
df.head()
```

```python
Z=linkage(df, 'complete')
plt.figure(figsize=(12, 5), dpi=200)
dendrogram(Z, labels=df.index, leaf_font_size=8)

plt.title('Hierarchical Clustering')
plt.xlabel('sample index')
plt.ylabel('distance')
plt.show()
```

```python
import scipy.cluster.hierarchy as shc

plt.figure(figsize=(12, 5), dpi=200)
clusters=shc.linkage(df, method='ward', metric='euclidean')
shc.dendrogram(Z=clusters)
# plt.axhline(y=125, color='r', linestyle='-')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

clustering_model=AgglomerativeClustering(n_clusters=20, affinity='euclidean', linkage='ward')
clustering_model.fit(df)
clustering_model.labels_
```
